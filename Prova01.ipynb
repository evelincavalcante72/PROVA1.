```python
# 1. Entrada: Perguntar ao usuário a idade e o valor total que ele tem na carteira
idade = int(input("Digite a sua idade: "))
dinheiro = float(input("Digite o valor total que você tem na carteira (R$): "))

# 2. Decisão
if idade < 16:
    print("Entrada permitida apenas com acompanhante.")
else:
    # Se a idade for 16 anos ou mais, verificar o dinheiro disponível
    if dinheiro >= 30.00:
        print("Ingresso comprado! Bom filme.")
    else:
        print("Saldo insuficiente para o ingresso.")

# 3. Finalização
print("Foi bom te atender!!")
```
